```python
# Remove Session and target columns from model inputs
feature_columns = [
    col for col in model_data_encoded.columns
    if col not in ["Session", "service_time_min"]
]

print("Number of current-patient features:",
      len(feature_columns))

print("\nFeatures:")
print(feature_columns)
```

```python
import numpy as np

SEQUENCE_LENGTH = 3

def create_sequences(data, feature_columns):
    X_history = []
    X_current = []
    y = []

    # Each session is processed independently
    for session_id, group in data.groupby("Session", sort=False):

        # Rows are already in chronological order
        group = group.reset_index(drop=True)

        # Convert current patient features into numerical values
        current_features = group[feature_columns].astype(float).values

        # Scaled actual consultation durations
        durations = group["service_time_scaled"].values

        # Start only after three completed consultations
        for i in range(SEQUENCE_LENGTH, len(group)):

            # Previous three completed consultation durations
            previous_durations = durations[
                i-SEQUENCE_LENGTH:i
            ]

            # Current patient's available features
            current_patient = current_features[i]

            # Current patient's duration is the target
            target = durations[i]

            X_history.append(previous_durations)
            X_current.append(current_patient)
            y.append(target)

    return (
        np.array(X_history, dtype=np.float32).reshape(-1, SEQUENCE_LENGTH, 1),
        np.array(X_current, dtype=np.float32),
        np.array(y, dtype=np.float32)
    )


X_train_history, X_train_current, y_train = create_sequences(
    train_data, feature_columns
)

X_val_history, X_val_current, y_val = create_sequences(
    validation_data, feature_columns
)

X_test_history, X_test_current, y_test = create_sequences(
    test_data, feature_columns
)

print("Training shapes:")
print("Historical input:", X_train_history.shape)
print("Current patient input:", X_train_current.shape)
print("Target:", y_train.shape)

print("\nValidation shapes:")
print("Historical input:", X_val_history.shape)
print("Current patient input:", X_val_current.shape)
print("Target:", y_val.shape)

print("\nTesting shapes:")
print("Historical input:", X_test_history.shape)
print("Current patient input:", X_test_current.shape)
print("Target:", y_test.shape)
```

```python
import tensorflow as tf

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, LSTM, Dense, Dropout,
    Concatenate
)
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

print("TensorFlow version:", tf.__version__)
```

```python
# Input 1: Previous three consultation durations
history_input = Input(
    shape=(SEQUENCE_LENGTH, 1),
    name="history_input"
)

# LSTM branch
history_features = LSTM(
    32,
    activation="tanh"
)(history_input)

history_features = Dropout(0.2)(history_features)


# Input 2: Current patient features
patient_input = Input(
    shape=(len(feature_columns),),
    name="patient_input"
)

# Dense branch
patient_features = Dense(
    32,
    activation="relu"
)(patient_input)


# Combine both branches
combined = Concatenate()([
    history_features,
    patient_features
])

combined = Dense(32, activation="relu")(combined)

combined = Dropout(0.2)(combined)

# One output: predicted scaled consultation duration
output = Dense(1, activation="linear")(combined)


# Create model
model = Model(
    inputs=[history_input, patient_input],
    outputs=output
)

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
def create_sequences_with_length(data, feature_columns, sequence_length):

    X_history = []
    X_current = []
    y = []

    for session_id, group in data.groupby("Session", sort=False):

        group = group.reset_index(drop=True)

        current_features = group[feature_columns].astype(float).values

        durations = group["service_time_scaled"].values

        for i in range(sequence_length, len(group)):

            X_history.append(
                durations[i-sequence_length:i]
            )

            X_current.append(current_features[i])

            # Actual duration in minutes
            y.append(group["service_time_min"].iloc[i])

    return (
        np.asarray(X_history, dtype=np.float32),
        np.asarray(X_current, dtype=np.float32),
        np.asarray(y, dtype=np.float32)
    )
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

sequence_lengths = [1, 2, 3, 5]

sequence_results = []

for seq_len in sequence_lengths:

    Xtr_hist, Xtr_current, ytr = create_sequences_with_length(
        train_data, feature_columns, seq_len
    )

    Xv_hist, Xv_current, yv = create_sequences_with_length(
        validation_data, feature_columns, seq_len
    )

    # Combine historical and current features
    Xtr = np.concatenate(
        [Xtr_current, Xtr_hist], axis=1
    )

    Xv = np.concatenate(
        [Xv_current, Xv_hist], axis=1
    )

    rf = RandomForestRegressor(
        n_estimators=300,
        max_depth=10,
        min_samples_split=5,
        min_samples_leaf=3,
        max_features=0.8,
        random_state=42,
        n_jobs=-1
    )

    rf.fit(Xtr, ytr)

    predictions = rf.predict(Xv)

    mae_value = mean_absolute_error(yv, predictions)

    rmse_value = np.sqrt(
        mean_squared_error(yv, predictions)
    )

    sequence_results.append({
        "Sequence Length": seq_len,
        "Training Samples": len(ytr),
        "Validation Samples": len(yv),
        "MAE": mae_value,
        "RMSE": rmse_value
    })

sequence_results_df = pd.DataFrame(sequence_results)

display(
    sequence_results_df.sort_values("MAE")
)
```

```python
SEQUENCE_LENGTH = 5

def create_lstm_sequences(data, feature_columns, sequence_length):

    X_history = []
    X_current = []
    y = []

    for session_id, group in data.groupby("Session", sort=False):

        group = group.reset_index(drop=True)

        current_features = group[feature_columns].astype(float).values
        durations = group["service_time_scaled"].values

        for i in range(sequence_length, len(group)):

            # Previous five completed consultations
            X_history.append(
                durations[i-sequence_length:i]
            )

            # Current patient's available features
            X_current.append(current_features[i])

            # Current consultation duration (scaled)
            y.append(durations[i])

    return (
        np.array(X_history, dtype=np.float32).reshape(
            -1, sequence_length, 1
        ),
        np.array(X_current, dtype=np.float32),
        np.array(y, dtype=np.float32)
    )


X_train_history_5, X_train_current_5, y_train_5 = (
    create_lstm_sequences(
        train_data, feature_columns, SEQUENCE_LENGTH
    )
)

X_val_history_5, X_val_current_5, y_val_5 = (
    create_lstm_sequences(
        validation_data, feature_columns, SEQUENCE_LENGTH
    )
)

print("Training shapes:")
print(X_train_history_5.shape)
print(X_train_current_5.shape)
print(y_train_5.shape)

print("\nValidation shapes:")
print(X_val_history_5.shape)
print(X_val_current_5.shape)
print(y_val_5.shape)
```

```python
def add_session_history(data):
    data = data.sort_values(
        ["Session", "StartTime_td"]
    ).copy()

    data["previous_session_avg"] = (
        data.groupby("Session")["service_time_min"]
        .transform(lambda x: x.shift(1).expanding().mean())
    )

    data["previous_session_std"] = (
        data.groupby("Session")["service_time_min"]
        .transform(lambda x: x.shift(1).expanding().std())
    )

    data["previous_session_count"] = (
        data.groupby("Session").cumcount()
    )

    return data


# Rebuild features using the original ordered dataset
train_sessions = train_data["Session"].unique()
val_sessions = validation_data["Session"].unique()
test_sessions = test_data["Session"].unique()

train_raw = add_session_history(
    hangu_ordered[hangu_ordered["Session"].isin(train_sessions)]
)

val_raw = add_session_history(
    hangu_ordered[hangu_ordered["Session"].isin(val_sessions)]
)

test_raw = add_session_history(
    hangu_ordered[hangu_ordered["Session"].isin(test_sessions)]
)

print("Training rows:", len(train_raw))
print("Validation rows:", len(val_raw))
print("Test rows:", len(test_raw))
```

```python
# Select historical features and retain their original row order
history_columns = [
    "previous_session_avg",
    "previous_session_std",
    "previous_session_count"
]

# Fill missing values using training data only
history_fill_avg = train_raw["previous_session_avg"].mean()

for dataset in [train_raw, val_raw, test_raw]:
    dataset["previous_session_avg"] = (
        dataset["previous_session_avg"].fillna(history_fill_avg)
    )
    dataset["previous_session_std"] = (
        dataset["previous_session_std"].fillna(0)
    )

# Attach features to encoded datasets
train_data_new = train_data.reset_index(drop=True).copy()
validation_data_new = validation_data.reset_index(drop=True).copy()
test_data_new = test_data.reset_index(drop=True).copy()

train_data_new[history_columns] = (
    train_raw[history_columns].reset_index(drop=True)
)

validation_data_new[history_columns] = (
    val_raw[history_columns].reset_index(drop=True)
)

test_data_new[history_columns] = (
    test_raw[history_columns].reset_index(drop=True)
)

print("Training shape:", train_data_new.shape)
print("Validation shape:", validation_data_new.shape)
print("Test shape:", test_data_new.shape)

print("\nMissing values:")
print(train_data_new[history_columns].isnull().sum())
```

```python
# Updated feature columns
new_feature_columns = feature_columns + [
    "previous_session_avg",
    "previous_session_std",
    "previous_session_count"
]

# Create five-step sequences
X_train_history_new, X_train_current_new, y_train_new = (
    create_lstm_sequences(
        train_data_new,
        new_feature_columns,
        5
    )
)

X_val_history_new, X_val_current_new, y_val_new = (
    create_lstm_sequences(
        validation_data_new,
        new_feature_columns,
        5
    )
)

X_test_history_new, X_test_current_new, y_test_new = (
    create_lstm_sequences(
        test_data_new,
        new_feature_columns,
        5
    )
)

print("Training shapes:")
print(X_train_history_new.shape)
print(X_train_current_new.shape)
print(y_train_new.shape)

print("\nValidation shapes:")
print(X_val_history_new.shape)
print(X_val_current_new.shape)
print(y_val_new.shape)

print("\nTest shapes:")
print(X_test_history_new.shape)
print(X_test_current_new.shape)
print(y_test_new.shape)
```

```python
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout, Concatenate
from tensorflow.keras.models import Model

# Input 1: Previous 5 consultation durations
history_input_new = Input(shape=(5, 1), name="history_input_new")

history_branch_new = LSTM(
    32,
    activation="tanh"
)(history_input_new)

history_branch_new = Dropout(0.2)(history_branch_new)

# Input 2: Current patient and historical session features
patient_input_new = Input(
    shape=(28,),
    name="patient_input_new"
)

patient_branch_new = Dense(
    32,
    activation="relu"
)(patient_input_new)

# Combine both inputs
combined_new = Concatenate()([
    history_branch_new,
    patient_branch_new
])

# Final prediction layers
combined_new = Dense(32, activation="relu")(combined_new)
combined_new = Dropout(0.2)(combined_new)

output_new = Dense(1, activation="linear")(combined_new)

# Create model
model_new = Model(
    inputs=[history_input_new, patient_input_new],
    outputs=output_new
)

model_new.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

model_new.summary()
```

```python
# Training callbacks
early_stop_new = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

reduce_lr_new = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_lr=0.00001
)

callbacks_new = [early_stop_new, reduce_lr_new]

# Train the improved LSTM
history_new = model_new.fit(
    [X_train_history_new, X_train_current_new],
    y_train_new,
    validation_data=(
        [X_val_history_new, X_val_current_new],
        y_val_new
    ),
    epochs=100,
    batch_size=32,
    callbacks=callbacks_new,
    verbose=1
)
```

```python
print("Epochs completed:", len(history_new.history["loss"]))
print("Best epoch:", np.argmin(history_new.history["val_loss"]) + 1)
print("Best validation loss:", min(history_new.history["val_loss"]))
print("Best validation MAE:", min(history_new.history["val_mae"]))
```

# LSTM-Based Consultation Duration Prediction

## 1. Objective

The objective of this experiment was to develop a deep learning model using Long Short-Term Memory (LSTM) networks to predict the consultation duration of patients in an outpatient department (OPD).

The model uses historical consultation durations and patient/session-related features to estimate the expected consultation time.

## 2. Dataset Description

The experiment uses the Hangu outpatient clinic dataset containing 6,637 consultation records across 381 sessions.

The dataset includes:
- Patient-related information
- Session and working-day information
- Consultation start time
- Consultation service duration

The target variable was created as:

`service_time_min = ServTime / 60`

The target represents consultation duration in minutes, not patient waiting time.

## 3. Data Preprocessing

The following preprocessing activities were performed:

1. Inspected missing values and duplicate records.
2. Converted consultation duration from seconds to minutes.
3. Sorted records chronologically within each session.
4. Encoded categorical variables using one-hot encoding.
5. Excluded patient IDs and unsuitable variables from model inputs.
6. Standardized the target variable using StandardScaler.

The dataset was divided by complete sessions:

| Dataset | Sessions | Records |
|---|---:|---:|
| Training | 266 | 4,603 |
| Validation | 57 | 1,028 |
| Testing | 58 | 1,006 |

## 4. LSTM Model Development

An LSTM architecture with two input branches was developed.

**Input 1: Historical consultation durations**

The model receives the previous five consultation durations as a sequential input.

**Input 2: Current patient and session features**

The second input contains 25 encoded patient/session features in the initial model.

The two branches are combined using a concatenation layer, followed by dense layers and a final linear output layer.

The model uses:
- LSTM with 32 units
- Dense layers with ReLU activation
- Dropout regularization
- Adam optimizer
- Mean Squared Error loss
- Mean Absolute Error metric

Early Stopping and ReduceLROnPlateau callbacks were used during training.

## 5. Feature Engineering Experiment

Three additional historical session features were introduced:

1. `previous_session_avg`
2. `previous_session_std`
3. `previous_session_count`

These features describe the consultation history available before the current patient.

The improved model therefore uses 28 current patient/session features alongside the five-step LSTM sequence.

Missing historical averages were replaced using the training-set historical average. Missing standard deviations were replaced with zero.

## 6. Model Evaluation

The following evaluation metrics were used:

- **MAE:** Mean Absolute Error
- **RMSE:** Root Mean Squared Error
- **R²:** Coefficient of Determination

### Final improved LSTM results

| Metric | Validation | Test |
|---|---:|---:|
| MAE | 4.33 minutes | 4.47 minutes |
| RMSE | 5.78 minutes | 5.77 minutes |
| R² | 0.0295 | 0.0197 |

## 7. Observations

The improved model achieved a small reduction in MAE and RMSE compared with the original five-step LSTM model.

However, the R² score remained low, indicating that the available features explain only a small proportion of the variation in consultation duration.

This suggests that additional relevant patient and consultation-related information may be necessary to improve individual predictions.

## 8. Conclusion

The experiment demonstrates the implementation of an LSTM-based regression model for consultation duration prediction.

Historical consultation sequences and session-level features were incorporated into the model. The results provide a baseline for further experimentation and integration into the proposed Smart OPD Queue Management System.

The model predicts consultation duration only. It does not independently predict the total patient waiting time.

```python
import numpy as np

def create_sequences(data, feature_columns, sequence_length=5):

    X = []
    y = []

    # Separate sequences by source sheet and arrival date
    for (source, date), group in data.groupby(
        ["source_sheet", "arrival_date"],
        sort=False
    ):

        group = group.sort_values("x_ArrivalDTTM")

        features = group[feature_columns].to_numpy(dtype=np.float32)
        target = group["actual_wait_min"].to_numpy(dtype=np.float32)

        # Create sequences
        for i in range(sequence_length - 1, len(group)):

            X.append(features[i - sequence_length + 1:i + 1])
            y.append(target[i])

    return (
        np.array(X, dtype=np.float32),
        np.array(y, dtype=np.float32)
    )
```

```python
# Generate sequences

X_train, y_train = create_sequences(
    train_data, selected_features, sequence_length=5
)

X_val, y_val = create_sequences(
    validation_data, selected_features, sequence_length=5
)

X_test, y_test = create_sequences(
    test_data, selected_features, sequence_length=5
)

# Display shapes

print("TRAINING:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nVALIDATION:")
print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("\nTESTING:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
feature_scaler = StandardScaler()

# Fit only on training records
feature_scaler.fit(train_data[selected_features])

print("Feature scaler fitted successfully!")
```

```python
# Get the original dimensions
train_shape = X_train.shape
val_shape = X_val.shape
test_shape = X_test.shape

# Reshape 3D arrays into 2D for scaling
X_train_2d = X_train.reshape(-1, len(selected_features))
X_val_2d = X_val.reshape(-1, len(selected_features))
X_test_2d = X_test.reshape(-1, len(selected_features))

# Apply the same training-fitted scaler
X_train_scaled = feature_scaler.transform(X_train_2d)
X_val_scaled = feature_scaler.transform(X_val_2d)
X_test_scaled = feature_scaler.transform(X_test_2d)

# Reshape back to 3D
X_train_scaled = X_train_scaled.reshape(train_shape)
X_val_scaled = X_val_scaled.reshape(val_shape)
X_test_scaled = X_test_scaled.reshape(test_shape)

print("Scaling completed!")
print("Training:", X_train_scaled.shape)
print("Validation:", X_val_scaled.shape)
print("Testing:", X_test_scaled.shape)
```

```python
print("Training feature mean:",
      X_train_scaled.mean())

print("Training feature standard deviation:",
      X_train_scaled.std())
```

```python
from sklearn.preprocessing import StandardScaler

# Create target scaler
target_scaler = StandardScaler()

# Fit only on training targets
y_train_scaled = target_scaler.fit_transform(
    y_train.reshape(-1, 1)
).flatten()

# Transform validation and testing targets
y_val_scaled = target_scaler.transform(
    y_val.reshape(-1, 1)
).flatten()

y_test_scaled = target_scaler.transform(
    y_test.reshape(-1, 1)
).flatten()

print("Target scaling completed!")

print("Training target mean:", y_train_scaled.mean())
print("Training target std:", y_train_scaled.std())

print("Validation shape:", y_val_scaled.shape)
print("Testing shape:", y_test_scaled.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    LSTM,
    Dense,
    Dropout
)
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

print("TensorFlow version:", tf.__version__)
```

```python
model = Sequential([
    Input(shape=(5, 21)),

    LSTM(64, activation="tanh"),

    Dropout(0.2),

    Dense(32, activation="relu"),

    Dense(1, activation="linear")
])

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

print("Model compiled successfully!")
```

```python
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=8,
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=0.00001
)

print("Callbacks configured successfully!")
```

```python
history = model.fit(
    X_train_scaled,
    y_train_scaled,
    validation_data=(X_val_scaled, y_val_scaled),
    epochs=50,
    batch_size=64,
    callbacks=[
        early_stopping,
        reduce_lr
    ],
    verbose=1
)
```

```python
# Find the epoch with the lowest validation loss

best_epoch = np.argmin(history.history["val_loss"]) + 1

best_val_loss = min(history.history["val_loss"])

print("Total epochs completed:", len(history.history["loss"]))
print("Best epoch:", best_epoch)
print("Best validation loss:", best_val_loss)

print(
    "Best validation MAE:",
    history.history["val_mae"][best_epoch - 1]
)
```

```python
# Plot training and validation loss

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("LSTM Training and Validation Loss")

plt.legend()
plt.grid(True)
plt.show()
```

```python
# Predict on test dataset

y_pred_scaled = model.predict(X_test_scaled)

# Convert predictions back into minutes

y_pred = target_scaler.inverse_transform(
    y_pred_scaled.reshape(-1, 1)
).flatten()

# Convert actual test values back into minutes

y_actual = target_scaler.inverse_transform(
    y_test_scaled.reshape(-1, 1)
).flatten()

print("Predictions generated successfully!")
print("Predictions shape:", y_pred.shape)
print("Actual values shape:", y_actual.shape)

# Display sample predictions

comparison = pd.DataFrame({
    "Actual_Wait_Min": y_actual[:10],
    "Predicted_Wait_Min": y_pred[:10]
})

display(comparison)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Calculate evaluation metrics

mae = mean_absolute_error(y_actual, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_actual, y_pred)
)

r2 = r2_score(y_actual, y_pred)

print("LSTM TEST RESULTS")
print("-------------------------")
print(f"MAE  : {mae:.2f} minutes")
print(f"RMSE : {rmse:.2f} minutes")
print(f"R2   : {r2:.4f}")
```
